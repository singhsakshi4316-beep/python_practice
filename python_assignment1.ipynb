```python
year = int(input("Enter a year: "))
if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("The year is a leap year.")
else:
    print("The year is not a leap year.")
```

```python
def is_prime(n):
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
        
    return True

for num in range(1, 31):
     if is_prime(num):
         print(num)
```

```python
def reverse_and_slice(text):
    reversed_text = ""
    for char in text:
        reversed_text = char + reversed_text

    half = text[:len(text)//2]
    return reversed_text, half

text=input("Enter a string: ")
reversed_text, half_text = reverse_and_slice(text)

print("Reversed string:", reversed_text)
print("First half of the string:", half_text)
```
